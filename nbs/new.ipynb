```python
from collections import Counter

def found(token, pair, pos):
  return token[pos:pos+2] == pair

def merge(token, pair, count, updates):
  (a, b) = pair 
  new_token, ab, i, n = [], a+b, 0, len(token)

  while(i < n - 1):
    if found(token, pair, i):
      new_token.append(ab)
      updates[(a, b)] -= count
      if i > 0:
        prev = token[i-1]
        updates[(prev, ab)] += count
        updates[(prev, a)] -= count
        
      # Consecutive pairs
      while(True): 
        if found(token, pair, i+2):
          new_token.append(ab)
          updates[(ab, ab)] += count
          updates[(b, a)] -= count
          i += 2
        # elif found(token, pair, i+1):
        #   updates[(ab, b)] += count
        #   updates[(b, a)] -= count
        #   i += 1
        else:
          break
      if i < n-2:
        nxt = token[i+2]
        updates[(ab, nxt)] += count
        updates[(b, nxt)] -= count
        i += 2
      i += 1
    else:
      new_token.append(token[i])
      i += 1

  return tuple(new_token)
  
```

```python
import regex as re
from collections import Counter

PAT = re.compile(r"""'(?:[sdmt]|ll|ve|re)| ?\p{L}+| ?\p{N}+| ?[^\s\p{L}\p{N}]+|\s+(?!\S)|\s+""")

def get_iter(text: str | list[str], pat: re.Pattern[str]):
  if isinstance(text, list):
    for line in text:
      yield from get_iter(line, pat)
  else:
    for match in pat.finditer(text):
      yield match.group()

# def list_iter(lst: list[str], pat: re.Pattern[str]):
#   for line in lst:
#     yield from string_iter(line, pat)

Counter(get_iter("one two three\n four", PAT))
Counter(get_iter([
  "one two three\n four", 
  "we are sitting on the floor"], PAT))
```
